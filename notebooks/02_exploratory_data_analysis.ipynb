# 02 — Exploratory Data Analysis & Business Insights
## EMIPredict AI Platform
This notebook explores financial distributions, credit ratings, living expense structures, and relationships with eligibility and maximum monthly EMI capacity.


```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import seaborn as sns

root_dir = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(root_dir) not in sys.path:
    sys.path.insert(0, str(root_dir))

from src.config import FIGURES_DIR
from src.data.load_data import load_raw_dataset
```

### 1. Load Dataset for Analysis


```python
df = load_raw_dataset()
FIGURES_DIR.mkdir(parents=True, exist_ok=True)
sns.set_theme(style='whitegrid', palette='muted')
```

### 2. Target Variable Analysis
We visualize the distribution of `emi_eligibility` and `max_monthly_emi`.


```python
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
sns.countplot(data=df, x='emi_eligibility', palette=['#00A896', '#F4A261', '#E63946'], ax=ax[0])
ax[0].set_title('EMI Eligibility Distribution', fontsize=12, fontweight='bold')

sns.histplot(df['max_monthly_emi'], kde=True, color='#028090', ax=ax[1])
ax[1].set_title('Max Monthly EMI Distribution (₹)', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig(FIGURES_DIR / 'target_distributions.png', dpi=300)
plt.show()
```

### 3. Income vs Credit Score & Eligibility
Analyzing how income bands and credit rating correlate with financial risk categories.


```python
fig, ax = plt.subplots(figsize=(8, 6))
sns.scatterplot(data=df.sample(2000, random_state=42), x='monthly_salary', y='credit_score', hue='emi_eligibility', alpha=0.7, palette=['#00A896', '#F4A261', '#E63946'], ax=ax)
ax.set_title('Monthly Salary vs Credit Score by Eligibility Tier', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig(FIGURES_DIR / 'salary_vs_credit_score.png', dpi=300)
plt.show()
```

### Summary of Business Insights
- Higher credit score (> 700) and substantial savings buffer strongly correlate with `Eligible` classification.
- Living expenses and existing loan EMIs significantly constrain `max_monthly_emi` capacity.
